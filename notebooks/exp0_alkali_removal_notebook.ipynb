# Calculating Theoretical capacity based on charge balance effects

```python
from mendeleev import element

formula = "Na0.66Li0.22Ti0.78O2"

formula_dict = analyzer.parse_formula(formula)

mass = 0

for atom in formula_dict.keys():
    weighted_mass = element(atom).mass * formula_dict[atom]
    mass += weighted_mass
    
print(mass)
```

n being 0.7 represents 0.7 of the Na being able to deintercalate here

```python
faradays_constant = 96485
n = 0.01
q = (n*faradays_constant)/(3.6*mass)

print(q)
```

this is the theoretical capacity normally calculated

```python
faradays_constant = 96485
n = 0.66
q = (n*faradays_constant)/(3.6*mass)

print(q)
```
